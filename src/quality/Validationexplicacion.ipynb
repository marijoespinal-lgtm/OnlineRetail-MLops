## Explicación de validation 

* Verificación de Columnas y Tipado de Fechas (pd.to_datetime): Garantiza el cumplimiento del esquema de datos antes de procesar el lote (Schema Enforcement). Convertir la columna InvoiceDate a tipo datetime64[ns] de forma explícita previene errores de ejecución (TypeError) durante las comparaciones temporales y asegura que el pipeline acepte formatos con horas y minutos.
* Tasa de Duplicados $< 5\%$ (dup_rate): Actúa como un umbral de tolerancia para el volumen de datos redundantes. Si la proporción de registros duplicados supera el $5\%$, indica una falla severa en el pipeline de ingesta o en los scripts de limpieza previa, lo cual detiene el flujo de datos para evitar sesgos en el entrenamiento de modelos.
* Evaluación de Integridad en CustomerID (isnull().sum()): Comprueba que no existan valores nulos en la identificación de clientes. En arquitecturas MLOps enfocadas en modelos de comportamiento (como RFM, Churn o recomendadores), la ausencia de un identificador único invalida la trazabilidad de los datos.
* Rango de Fechas Esperado (2009–2011): Valida la ventana temporal del negocio para detectar data drift o registros corruptos del sistema fuente. El límite superior extendido a "2011-12-31 23:59:59" asegura que las transacciones realizadas al final del día del último año no sean descartadas por truncamiento de hora.
* Rango Válido de Precios ($0 < \text{UnitPrice} < 10\,000$): Confirma la consistencia de las variables financieras. Garantiza que no hayan ingresado precios negativos/nulos (que distorsionan métricas como el valor promedio de pedido) ni valores atípicamente altos que afecten la convergencia de modelos basados en distancia o regresión.
* Rango Válido de Cantidades ($0 < \text{Quantity} < 10\,000$): Asegura que las unidades vendidas se mantengan dentro de un comportamiento comercial coherente para clientes minoristas. Filtra posibles anomalías o errores de digitación en el sistema origen antes de que pasen a la etapa de generación de características (Feature Engineering).
