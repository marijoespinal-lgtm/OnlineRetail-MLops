## Explicación del proceso de contaminación

* Muestreo Aislado (sample): Se toma una submuestra representativa del dataset limpio sin modificar la fuente original, garantizando la idempotencia y seguridad de los datos originales en data/clean/.
* Casting de Tipos (astype(object)): Se cambia el tipo de dato de la columna UnitPrice de flotante a objeto. Esto permite inyectar cadenas de texto en columnas numéricas para probar que la validación detecte fallos de tipado estricto (Type Checking).
* Inyección de Precios Inválidos: Se insertan precios negativos (-15.50) y cadenas incoherentes ("PRECIO_INVALIDO") para simular errores de captura o corrupción en la base de datos origen.Inyección de Cantidades Negativas: Se fuerzan valores de Quantity menores a cero (-50) para evaluar si los Quality Gates detectan transacciones inconsistentes.
* Inyección de Fechas Anómalas (Data Drift Temporal): Se reemplazan fechas por valores fuera de la ventana operativa proyectada (año 2005), simulando desajustes temporales o registros desfasados.Inyección de Nulos en Variables Críticas (np.nan): Se remueven identificadores de cliente (CustomerID) para verificar el bloqueo ante pérdida de trazabilidad de usuarios.
* Inyección Masiva de Duplicados: Se replica un bloque de filas dentro del mismo archivo para elevar la tasa de redundancia por encima del umbral permitido ($>15\%$), activando las alertas por duplicación excesiva.3. Resultado en la Arquitectura MLOpsEl archivo contaminado se almacena de forma aislada en data/corrupted/online_retail_corrupted.csv. Al ejecutar la rutina de validación sobre este archivo, el sistema interrumpe inmediatamente el pipeline (Fail-Fast Principle), previniendo la degradación del modelo y garantizando que ningún dato fuera de norma sea procesado.
